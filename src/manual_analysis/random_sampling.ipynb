```python
directory_path = '../../manual_analysis/python/'
path = '../../manual_analysis/'
sample_path = '../../manual_analysis/samples/'
# files = ['java','cpp','js','python']
```

```python
# read files in directory path
files = os.listdir(directory_path)
df = pd.DataFrame()

# merge all files into one dataframe in the directory
for file in files:
    df2 = pd.read_csv(directory_path+file)
    df = pd.concat([df,df2],ignore_index=True)

print(df.shape)
df.to_csv(path+'python.csv',index=False)
```

```python
subject_systems = ['java','cpp','js','python']
for name in subject_systems:
    df = pd.read_csv(path + name + '.csv')
    # randomly sample 244 rows
    df = df.sample(n=244)
    # save the sampled data
    df.to_csv(sample_path + name + '.csv', index=False)
```
